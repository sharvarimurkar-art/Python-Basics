10.What is the difference between implicit and explicit type casting?
  -Implicit Type Casting
  Automatic: Performed by the compiler or interpreter without the need for explicit instructions from the programmer.
  Widening Conversion: Often occurs when converting a "smaller" data type to a "larger" one, such as an integer to a float, to preserve data integrity.
  No Data Loss: Tends to avoid data loss, as the conversion is into a type that can hold the original value.
  Examples: Assigning an integer to a float variable or adding an integer to a double in C or other languages.
  -Explicit Type Casting
  Manual: Requires the programmer to use specific syntax to force a conversion.
  Narrowing or Specific Conversion: Can be used to convert data to a smaller type or to a different, but related, type based on the programmer's intent.
  Programmer Control: Gives the programmer direct control over the conversion process, allowing for specific type manipulations.
  Examples: Converting a string to an integer using int("123") in Python, or explicitly casting a double to an int by truncating the decimal part.

12.How does the elif statement work?
  -how it works:
  Initial if check: The program first evaluates the condition within the if statement.
  elif evaluation (if if is false): If the if condition is False, the program proceeds to the first elif statement (if present) and evaluates its condition.
  Subsequent elif evaluations: If the current elif condition is False, the program moves to the next elif statement in the sequence and evaluates its condition. This process continues for all subsequent elif statements until a True condition is found.
  Code execution: When an if or elif condition evaluates to True, the block of code associated with that condition is executed, and the entire if-elif-else block is exited. No further elif or else conditions are checked.
  else block (if all preceding conditions are false): If all if and elif conditions in the sequence evaluate to False, the code block within the else statement (if present) is executed.

7.Write a program that performs basic arithmetic operations on two user-input numbers.

```python
def perform_arithmetic_operations():
    try:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))
    except ValueError:
        print("Invalid input. Please enter valid numbers.")
        return

    operator = input("Enter an operator (+, -, *, /): ")

    if operator == '+':
        result = num1 + num2
        print(f"Result: {num1} + {num2} = {result}")
    elif operator == '-':
        result = num1 - num2
        print(f"Result: {num1} - {num2} = {result}")
    elif operator == '*':
        result = num1 * num2
        print(f"Result: {num1} * {num2} = {result}")
    elif operator == '/':
        if num2 != 0:
            result = num1 / num2
            print(f"Result: {num1} / {num2} = {result}")
        else:
            print("Error: Division by zero is not allowed.")
    else:
        print("Invalid operator. Please use +, -, *, or /.")

if __name__ == "__main__":
    perform_arithmetic_operations()
```

9.Write a Python program to convert user input from string to integer, float, and boolean types

```python
def convert_user_input():
    """
    Takes user input as a string and attempts to convert it to an integer,
    a float, and a boolean, printing the results and handling errors.
    """
    user_input_str = input("Enter a value to convert: ")

    try:
        converted_int = int(user_input_str)
        print(f"Converted to Integer: {converted_int} (Type: {type(converted_int)})")
    except ValueError:
        print(f"Could not convert '{user_input_str}' to an Integer.")

    try:
        converted_float = float(user_input_str)
        print(f"Converted to Float: {converted_float} (Type: {type(converted_float)})")
    except ValueError:
        print(f"Could not convert '{user_input_str}' to a Float.")

    user_input_lower = user_input_str.strip().lower()
    if user_input_lower == 'true':
        converted_bool = True
        print(f"Converted to Boolean: {converted_bool} (Type: {type(converted_bool)})")
    elif user_input_lower == 'false':
        converted_bool = False
        print(f"Converted to Boolean: {converted_bool} (Type: {type(converted_bool)})")
    else:
        print(f"'{user_input_str}' is not a standard boolean string ('True' or 'False').")

convert_user_input()
```

14.Write a program to reverse a string using a while loop.

```python
def reverse_string_while(input_string):
  """
  Reverses a given string using a while loop.

  Args:
    input_string: The string to be reversed.

  Returns:
    The reversed string.
  """
  reversed_str = ""
  index = len(input_string) - 1  # Start from the last character's index

  while index >= 0:
    reversed_str += input_string[index]
    index -= 1  # Move to the previous character
  return reversed_str

original_string = "Hello, World!"
reversed_result = reverse_string_while(original_string)
print(f"Original string: {original_string}")
print(f"Reversed string: {reversed_result}")
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop

```python
def calculate_factorial_while(number):
    """
    Calculates the factorial of a non-negative integer using a while loop.

    Args:
        number: The non-negative integer for which to calculate the factorial.

    Returns:
        The factorial of the number, or an error message if the input is negative.
    """
    if number < 0:
        return "Factorial is not defined for negative numbers."
    elif number == 0:
        return 1
    else:
        factorial = 1
        current_num = 1
        while current_num <= number:
            factorial *= current_num
            current_num += 1
        return factorial

try:
    num_input = int(input("Enter a non-negative integer to calculate its factorial: "))
    result = calculate_factorial_while(num_input)
    print(f"The factorial of {num_input} is: {result}")
except ValueError:
    print("Invalid input. Please enter an integer.")
```
